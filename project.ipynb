```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error

data = pd.read_csv("students_data.csv")

x = data[["Hours"]]     # Double bracket for 2D input
y = data["Score"]       # Target column

model = LinearRegression()

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

model.fit(x_train, y_train)

predicted_score = model.predict(x_test)

mae = mean_absolute_error(y_test, predicted_score)
mse = mean_squared_error(y_test, predicted_score)
rmse = root_mean_squared_error(y_test, predicted_score)

print("Mean Absolute Error is: ", mae)
print("Mean Squared Error is: ", mse)
print("Root Mean Squared Error is: ", rmse)


input_hours = float(input("Enter the Number of Hours: "))
new_predicted_score = model.predict([[input_hours]])
print(f"The Scores predicted based on {input_hours} hours are: ", new_predicted_score)
```
